```python
import json


# Load the data ~ let's write a function which can get the data loaded

filePath = "/Users/devanshshah/Developer/PrimeLab/Jupyter_Lab_Files/Data/movie_data.json"

def load_data(fileName):
    with open(fileName,"r") as f:
        data = json.load(f)

    return data
```

```python
data = load_data(filePath)
```

```python
# Clean & Structure The Data

def clean_data(data):
    handle_missing_data(data)
    clean_rating(data)
    remove_duplicates(data)
```

```python
def clean_rating(data):
    text_to_num = { "zero" : 0, "one": 1.0, "two": 2.0, "three": 3.0, "four": 4.0, "five": 5.0, "six": 6.0, "seven": 7.0, "eight": 8.0, "nine": 9.0, "ten": 10.0 }
    
    for movie in data:
        
        # clean the rating part
        rawRating = movie["rating"]
        
        if isinstance(rawRating,str):
            rawRating = rawRating.strip().lower()

            if rawRating in text_to_num:
                rawRating = text_to_num[rawRating]

        try:
            if rawRating is None:
                continue
            rawRating = float(rawRating)
        
        except ValueError:
            rawRating = None
                
        movie["rating"] = rawRating
```

```python
def handle_missing_data(data):
    
    required_fields = ["title","director","year","genre","rating","duration_minutes"]
    
    for movie in data:
        for field in required_fields:
            if field not in movie or movie[field] is None:
                movie[field] = None                
```

```python
def remove_duplicates(data):
    already_seen_movies = set()
    unique_movies = []

    for movie in data:
        movie_key = tuple(movie.items())

        if movie_key not in already_seen_movies:
            already_seen_movies.add(movie_key)
            unique_movies.append(movie)

    data.clear()
    data.extend(unique_movies)
```

```python
clean_data(data)
```

```python
# 3. Save cleaned data

with open(filePath, "w") as f:

    json.dump(data, f, indent=2)
```

```python
def getInsights(data):
    total_rating = 0
    valid_ratings = 0

    for movie in data:
        rating = movie["rating"]

        if rating is None:
            continue

        total_rating += rating
        valid_ratings += 1

    print(f"Average Rating = {total_rating / valid_ratings}")
```

```python
getInsights(data)
```
