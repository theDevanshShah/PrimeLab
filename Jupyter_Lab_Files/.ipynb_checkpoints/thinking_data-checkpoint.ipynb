```python
import json


# Load the data ~ let's write a function which can get the data loaded

def load_data(fileName):
    with open(fileName,"r") as f:
        data = json.load(f)

    return data
```

```python
def clean_rating(data):
text_to_num = { "one": 1.0, "two": 2.0, "three": 3.0, "four": 4.0, "five": 5.0, "six": 6.0, "seven": 7.0, "eight": 8.0, "nine": 9.0, "ten": 10.0 }
    
    for movie in data:
        
        # clean the rating part
        rawRating = movie["rating"]
        
        if isinstance(rawRating,str):
            rawRating = rawRating.strip().lower()

            if rawRating in text_to_num:
                rawRating = text_to_num[rawRating]

        try:
            rawRating = float(rawRating)
        except ValueError:
            rawRating = None
                
        movie["rating"] = rawRating
```

```python
# 3. Save cleaned data

with open("Jupyter_Lab_Files/Data/movie_data.json", "w") as f:

    json.dump(data, f, indent=2)
```
